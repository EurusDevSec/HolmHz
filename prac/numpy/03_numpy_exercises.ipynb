## **Bai1: Normalization**

```python
import numpy as np

data = np.random.randint(0,1000, size=10)
print(data)
```

```python
X_min = np.min(data)
X_max = np.max(data)
data_new = (data - X_min)/ (X_max - X_min)
print(data_new)
```
